```python
import pandas as pd
import glob 
pd.set_option('future.no_silent_downcasting', True)
```

```python
arquivos = glob.glob("*.xls")
```

```python
lista = []
```

```python
def limpar(parametro):
    df = pd.read_excel(parametro, header=None)
    df.dropna(how="all", inplace=True)
    df.reset_index(drop=True, inplace=True)
    df = df.iloc[3:]
    df.reset_index(drop=True, inplace=True)
    df.columns = df.iloc[0]
    df = df[1:]
    df.replace("-", pd.NA, inplace=True)
    df = df.infer_objects(copy=False)
    return df
```

```python
for arquivo in arquivos:
    df = limpar(arquivo)
    partes = arquivo.split("_")
    data = partes[1].replace("Subitem.xls","")

    df["Ano"] = data[:4]
    df["Mes"] = data[4:]

    lista.append(df)
```

```python
dataframe = pd.concat(lista, ignore_index=True)
```

```python
dataframe = dataframe.rename(columns={dataframe.columns[0]: "Subitem"})
```

```python
melt = dataframe.melt(
    id_vars = ["Subitem", "Ano", "Mes"],
    var_name = "Capital",
    value_name = "Valor"
)
```

```python
melt["Data"] = pd.to_datetime(melt["Ano"] + "-" + melt["Mes"] + "-01")
```

```python
melt = melt.drop(columns=["Ano", "Mes"])
```

```python
melt["Valor"] = pd.to_numeric(melt["Valor"])
```

```python
melt.duplicated().sum()
len(melt["Subitem"].unique())
melt.groupby("Data").size()
```

```python
melt.to_excel("IPCA2025.xlsx", index=False)
```
